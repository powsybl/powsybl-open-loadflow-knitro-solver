```python
# Packages
from slack_viz_utils import nad_explorer_with_slack,compute_slack_info
import pandas as pd
import pypowsybl as pp
import pypowsybl.network as pn
```

Load Data from JAVA, both the iidm network and the csv with the slacks results 

```python
network = pn.load('./Outputs/ieee30-voltage-perturbation.xiidm')
data = pd.read_csv('test_V_IEEE/ieee30/ieee30_voltage-perturbation.csv', sep=";")
```

```python
# data sturcturing, join slack data and the network information for visualisation
slack_info = compute_slack_info(network, data)
```

```python
explorer = nad_explorer_with_slack(network, slack_info=slack_info)
explorer
```

```python
# Build a bus-level dataframe with slack type and associated color
_default_color_map = {'P': 'red', 'Q': 'blue', 'V': 'green'}
slack_type_colors = _default_color_map

bus_node_style_df = (
    slack_info[['bus_id', 'type']]
    .copy()
    .rename(columns={
        'bus_id': 'id',
        'type': 'slack_type'
    })
)

# bus_descriptions_df['slack_type'] = bus_descriptions_df['slack_type'].fillna('none')
bus_node_style_df['fill'] = bus_node_style_df['slack_type'].map(slack_type_colors).fillna('gray')
bus_node_style_df = bus_node_style_df.set_index('id').rename_axis('id').drop(columns=['slack_type'])
```

```python
diagram_profile=pn.NadProfile( bus_node_styles=bus_node_style_df)#, slack_info =enhanced_bus_descriptions_df )#,bus_descriptions_df_descriptions=enhanced_bus_descriptions_df)#, bus_descriptions=enhanced_bus_descriptions_df, vl_descriptions=enhanced_vl_descriptions_df)
colored_explorer = nad_explorer_with_slack(network, nad_profile =diagram_profile, slack_info=slack_info )#(network, slack_info=slack_info, bus_node_styles= bus_node_style_df )
colored_explorer
```
